```python
def iqr_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return (data[column] < lower) | (data[column] > upper)


height_outliers = iqr_outliers(df, "height")
weight_outliers = iqr_outliers(df, "weight")

print("Height outliers:", height_outliers.sum())
print("Weight outliers:", weight_outliers.sum())
```

# Machine Learning-Based Cardiovascular Risk Prediction

## Objective

The objective of this section is to develop and compare machine learning models for predicting cardiovascular disease from routinely collected health and demographic features.

```python
# ==========================================
# STEP 1: Feature and Target Preparation
# ==========================================

from sklearn.model_selection import train_test_split

# Columns that are NOT model inputs:
#   id        -> identifier only
#   age       -> age in days; replaced by age_years
#   age_group -> derived from age_years (redundant); kept only for EDA and subgroup analysis
#   cardio    -> target
DROP_COLS = ["id", "age", "age_group", "cardio"]

X = df.drop(columns=DROP_COLS)
y = df["cardio"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget proportions:")
print((y.value_counts(normalize=True) * 100).round(2))

# ==========================================
# Train-Test Split
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTrain set:", X_train.shape)
print("Test set:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# ==========================================
# STEP 2: Preprocessing Pipeline
# ==========================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Numeric features
numeric_features = [
    "age_years",
    "height",
    "weight",
    "ap_hi",
    "ap_lo"
]

# Categorical features
categorical_features = [
    "gender",
    "cholesterol",
    "gluc",
    "smoke",
    "alco",
    "active"
]

# Preprocessing for numerical variables
numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

# Preprocessing for categorical variables.
# drop="if_binary" keeps ONE column for 0/1 features (e.g. smoke_1) instead of
# two mirror-image columns (smoke_0 and smoke_1).
categorical_transformer = Pipeline(
    steps=[
        ("onehot", OneHotEncoder(
            drop="if_binary",
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]
)

# Combine preprocessing (X already contains only the columns listed here)
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

assert set(numeric_features + categorical_features) == set(X.columns), "Feature lists do not match X"

print("Preprocessing pipeline created successfully.")
print("\nNumeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
# ==========================================
# STEP 3: Logistic Regression Baseline
# ==========================================

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# Create complete pipeline
logistic_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]
)

# Train the model
logistic_pipeline.fit(X_train, y_train)

# Predictions
y_pred_lr = logistic_pipeline.predict(X_test)
y_prob_lr = logistic_pipeline.predict_proba(X_test)[:, 1]

# Evaluation metrics
lr_accuracy = accuracy_score(y_test, y_pred_lr)
lr_precision = precision_score(y_test, y_pred_lr)
lr_recall = recall_score(y_test, y_pred_lr)
lr_f1 = f1_score(y_test, y_pred_lr)
lr_roc_auc = roc_auc_score(y_test, y_prob_lr)
lr_pr_auc = average_precision_score(y_test, y_prob_lr)

print("=== Logistic Regression Results ===")
print(f"Accuracy : {lr_accuracy:.4f}")
print(f"Precision: {lr_precision:.4f}")
print(f"Recall   : {lr_recall:.4f}")
print(f"F1 Score : {lr_f1:.4f}")
print(f"ROC-AUC  : {lr_roc_auc:.4f}")
print(f"PR-AUC   : {lr_pr_auc:.4f}")

print("\n=== Confusion Matrix ===")
print(confusion_matrix(y_test, y_pred_lr))

print("\n=== Classification Report ===")
print(classification_report(y_test, y_pred_lr))
```

```python
# ==========================================
# STEP 4: Random Forest Model
# ==========================================

from sklearn.ensemble import RandomForestClassifier

# Create Random Forest pipeline
rf_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=300,
            max_depth=None,
            min_samples_split=5,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

# Train the model
rf_pipeline.fit(X_train, y_train)

# Predictions
y_pred_rf = rf_pipeline.predict(X_test)
y_prob_rf = rf_pipeline.predict_proba(X_test)[:, 1]

# Evaluation
rf_accuracy = accuracy_score(y_test, y_pred_rf)
rf_precision = precision_score(y_test, y_pred_rf)
rf_recall = recall_score(y_test, y_pred_rf)
rf_f1 = f1_score(y_test, y_pred_rf)
rf_roc_auc = roc_auc_score(y_test, y_prob_rf)
rf_pr_auc = average_precision_score(y_test, y_prob_rf)

print("=== Random Forest Results ===")
print(f"Accuracy : {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall   : {rf_recall:.4f}")
print(f"F1 Score : {rf_f1:.4f}")
print(f"ROC-AUC  : {rf_roc_auc:.4f}")
print(f"PR-AUC   : {rf_pr_auc:.4f}")

print("\n=== Confusion Matrix ===")
print(confusion_matrix(y_test, y_pred_rf))

print("\n=== Classification Report ===")
print(classification_report(y_test, y_pred_rf))
```

```python
# ==========================================
# STEP 5: XGBoost Model
# ==========================================

from xgboost import XGBClassifier

# Create XGBoost pipeline
xgb_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", XGBClassifier(
            n_estimators=300,
            max_depth=5,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42,
            n_jobs=-1
        ))
    ]
)

# Train the model
xgb_pipeline.fit(X_train, y_train)

# Predictions
y_pred_xgb = xgb_pipeline.predict(X_test)
y_prob_xgb = xgb_pipeline.predict_proba(X_test)[:, 1]

# Evaluation
xgb_accuracy = accuracy_score(y_test, y_pred_xgb)
xgb_precision = precision_score(y_test, y_pred_xgb)
xgb_recall = recall_score(y_test, y_pred_xgb)
xgb_f1 = f1_score(y_test, y_pred_xgb)
xgb_roc_auc = roc_auc_score(y_test, y_prob_xgb)
xgb_pr_auc = average_precision_score(y_test, y_prob_xgb)

print("=== XGBoost Results ===")
print(f"Accuracy : {xgb_accuracy:.4f}")
print(f"Precision: {xgb_precision:.4f}")
print(f"Recall   : {xgb_recall:.4f}")
print(f"F1 Score : {xgb_f1:.4f}")
print(f"ROC-AUC  : {xgb_roc_auc:.4f}")
print(f"PR-AUC   : {xgb_pr_auc:.4f}")

print("\n=== Confusion Matrix ===")
print(confusion_matrix(y_test, y_pred_xgb))

print("\n=== Classification Report ===")
print(classification_report(y_test, y_pred_xgb))
```

## Hyperparameter Tuning

The default XGBoost settings were chosen by hand. Here they are tuned with a randomized search (3-fold cross-validation, ROC-AUC) **on the training set only**, so the test set stays untouched. The tuned model is only used if it beats the default configuration on the same cross-validation folds.

```python
# ==========================================
# STEP 7: Hyperparameter Tuning (XGBoost)
# ==========================================

from scipy.stats import randint, uniform
from sklearn.base import clone
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold, cross_val_score

def evaluate_probs(y_true, prob, thr=0.5):
    """Standard metric set for predicted probabilities at a given threshold."""
    pred = (prob >= thr).astype(int)
    return {
        "Accuracy": accuracy_score(y_true, pred),
        "Precision": precision_score(y_true, pred),
        "Recall": recall_score(y_true, pred),
        "F1 Score": f1_score(y_true, pred),
        "ROC-AUC": roc_auc_score(y_true, prob),
        "PR-AUC": average_precision_score(y_true, prob),
    }

cv3 = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

search_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", XGBClassifier(
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42,
            n_jobs=1            # parallelism is handled by the search itself
        ))
    ]
)

param_distributions = {
    "model__n_estimators": randint(200, 800),
    "model__max_depth": randint(3, 9),
    "model__learning_rate": uniform(0.01, 0.14),      # 0.01 - 0.15
    "model__subsample": uniform(0.6, 0.4),            # 0.6 - 1.0
    "model__colsample_bytree": uniform(0.6, 0.4),     # 0.6 - 1.0
    "model__min_child_weight": randint(1, 10),
    "model__gamma": uniform(0, 0.5),
    "model__reg_lambda": uniform(0.5, 4.5),
    "model__reg_alpha": uniform(0, 1),
}

search = RandomizedSearchCV(
    search_pipeline,
    param_distributions=param_distributions,
    n_iter=30,
    scoring="roc_auc",
    cv=cv3,
    random_state=42,
    n_jobs=-1,
    refit=True,
    verbose=1
)
search.fit(X_train, y_train)

# Default configuration evaluated on exactly the same folds
baseline_cv_auc = cross_val_score(
    clone(xgb_pipeline), X_train, y_train, cv=cv3, scoring="roc_auc"
).mean()

print(f"\nDefault XGBoost CV ROC-AUC : {baseline_cv_auc:.4f}")
print(f"Tuned XGBoost   CV ROC-AUC : {search.best_score_:.4f}")
print("\nBest parameters:")
for k, v in search.best_params_.items():
    print(f"  {k.replace('model__', ''):<18} {v:.4f}" if isinstance(v, float) else f"  {k.replace('model__', ''):<18} {v}")

# Top 5 configurations
search_results = (
    pd.DataFrame(search.cv_results_)
    .sort_values("rank_test_score")
    .head(5)[["rank_test_score", "mean_test_score", "std_test_score"]]
    .round(4)
)
display(search_results)

# Tuned model on the untouched test set
tuned_pipeline = search.best_estimator_
tuned_pipeline.set_params(model__n_jobs=-1)
y_prob_tuned = tuned_pipeline.predict_proba(X_test)[:, 1]

tuned_row = {"Model": "XGBoost (tuned)"}
tuned_row.update({k: round(v * 100, 2) for k, v in evaluate_probs(y_test, y_prob_tuned).items()})

model_results_all = pd.concat(
    [model_results_display, pd.DataFrame([tuned_row])],
    ignore_index=True
)
print("\n=== Model Comparison on Test Set (%) ===")
display(model_results_all)

# Choose the final XGBoost variant using the CROSS-VALIDATION score (not the test set)
if search.best_score_ > baseline_cv_auc:
    final_pipeline, final_name = tuned_pipeline, "XGBoost (tuned)"
else:
    final_pipeline, final_name = xgb_pipeline, "XGBoost (default)"

y_prob_final = final_pipeline.predict_proba(X_test)[:, 1]
print(f"\nSelected model: {final_name}")
```

## Are the Model Differences Real?

The test set has 13,743 records, so accuracy has a standard error of roughly 0.4 percentage points. Differences between the models can therefore be within noise. Two checks:

1. **5-fold cross-validation** (mean ± standard deviation) for the three baseline models.
2. **Bootstrap confidence intervals** for ROC-AUC on the test set, including paired differences between models.

```python
# ==========================================
# STEP 8: 5-Fold Cross-Validation (mean ± std)
# ==========================================

from sklearn.model_selection import cross_validate

cv5 = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_scoring = {
    "Accuracy": "accuracy",
    "Precision": "precision",
    "Recall": "recall",
    "F1 Score": "f1",
    "ROC-AUC": "roc_auc",
    "PR-AUC": "average_precision",
}

cv_models = {
    "Logistic Regression": logistic_pipeline,
    "Random Forest": rf_pipeline,
    "XGBoost (default)": xgb_pipeline,
}

cv_raw = {}
cv_display_rows = []
cv_records = []

for model_name, pipe in cv_models.items():
    scores = cross_validate(pipe, X, y, cv=cv5, scoring=cv_scoring, n_jobs=1)
    cv_raw[model_name] = scores

    display_row = {"Model": model_name}
    for metric in cv_scoring:
        vals = scores[f"test_{metric}"] * 100
        display_row[metric] = f"{vals.mean():.2f} ± {vals.std(ddof=1):.2f}"
        cv_records.append({
            "Model": model_name,
            "Metric": metric,
            "Mean": float(vals.mean()),
            "Std": float(vals.std(ddof=1)),
        })
    cv_display_rows.append(display_row)

cv_table = pd.DataFrame(cv_display_rows)
print("=== 5-Fold Cross-Validation (%, mean ± std over folds) ===")
display(cv_table)

# Paired, fold-by-fold ROC-AUC difference (same folds for every model)
for other in ["Logistic Regression", "Random Forest"]:
    diff = (cv_raw["XGBoost (default)"]["test_ROC-AUC"] - cv_raw[other]["test_ROC-AUC"]) * 100
    print(f"XGBoost - {other:<20} ROC-AUC per fold: mean {diff.mean():+.2f} pts, "
          f"min {diff.min():+.2f}, max {diff.max():+.2f}")
```

```python
# ==========================================
# STEP 9: Bootstrap Confidence Intervals (ROC-AUC, test set)
# ==========================================

N_BOOT = 1000
rng = np.random.default_rng(42)
y_true_arr = np.asarray(y_test)
n_test = len(y_true_arr)

boot_models = {
    "Logistic Regression": y_prob_lr,
    "Random Forest": y_prob_rf,
    "XGBoost (default)": y_prob_xgb,
    "XGBoost (tuned)": y_prob_tuned,
}

boot_auc = {name: np.empty(N_BOOT) for name in boot_models}

for b in range(N_BOOT):
    idx = rng.integers(0, n_test, n_test)   # same resample for every model -> paired comparison
    y_b = y_true_arr[idx]
    for name, prob in boot_models.items():
        boot_auc[name][b] = roc_auc_score(y_b, prob[idx])

boot_records = []
for name, prob in boot_models.items():
    lo, hi = np.percentile(boot_auc[name], [2.5, 97.5])
    boot_records.append({
        "Model": name,
        "ROC-AUC": round(roc_auc_score(y_true_arr, prob), 4),
        "95% CI low": round(lo, 4),
        "95% CI high": round(hi, 4),
    })

boot_table = pd.DataFrame(boot_records)
print(f"=== ROC-AUC with 95% bootstrap CI ({N_BOOT} resamples) ===")
display(boot_table)

# Paired differences
print("\n=== Paired ROC-AUC differences ===")
paired_records = []
for a, b_name in [("XGBoost (default)", "Logistic Regression"),
                  ("XGBoost (default)", "Random Forest"),
                  ("XGBoost (tuned)", "XGBoost (default)")]:
    d = boot_auc[a] - boot_auc[b_name]
    lo, hi = np.percentile(d, [2.5, 97.5])
    verdict = "difference is distinguishable from 0" if (lo > 0 or hi < 0) else "CI includes 0 (not distinguishable)"
    paired_records.append({"Comparison": f"{a} - {b_name}", "Mean diff": round(d.mean(), 4),
                           "95% CI low": round(lo, 4), "95% CI high": round(hi, 4), "Verdict": verdict})
paired_table = pd.DataFrame(paired_records)
display(paired_table)

# Accuracy standard error on the test set
acc = accuracy_score(y_test, (y_prob_final >= 0.5).astype(int))
se = np.sqrt(acc * (1 - acc) / n_test)
print(f"\nAccuracy of the selected model: {acc:.4f} (standard error ≈ ±{se:.4f})")
```

## Probability Calibration

The app shows a risk **probability**, so it matters whether "80%" really means that about 80 out of 100 similar patients have the disease. The reliability curve compares predicted probability with the observed rate. If the model is not well calibrated, an isotonic calibration layer is fitted on the training set and used only if it lowers the Brier score on the test set.

```python
# ==========================================
# STEP 14: Calibration Check
# ==========================================

from sklearn.calibration import calibration_curve, CalibratedClassifierCV
from sklearn.metrics import brier_score_loss

def expected_calibration_error(y_true, prob, n_bins=10):
    y_true = np.asarray(y_true)
    edges = np.linspace(0, 1, n_bins + 1)
    ids = np.clip(np.digitize(prob, edges[1:-1]), 0, n_bins - 1)
    ece = 0.0
    for b in range(n_bins):
        m = ids == b
        if m.any():
            ece += m.mean() * abs(y_true[m].mean() - prob[m].mean())
    return ece

# Isotonic calibration on top of the selected pipeline (fitted on the training set only)
calibrated_model = CalibratedClassifierCV(
    estimator=clone(final_pipeline),
    method="isotonic",
    cv=5
)
calibrated_model.fit(X_train, y_train)
y_prob_cal = calibrated_model.predict_proba(X_test)[:, 1]

calib_rows = []
plt.figure(figsize=(8, 7))
for label, prob in [(final_name, y_prob_final), (f"{final_name} + isotonic", y_prob_cal)]:
    frac_pos, mean_pred = calibration_curve(y_test, prob, n_bins=10, strategy="quantile")
    plt.plot(mean_pred, frac_pos, marker="o", label=label)
    calib_rows.append({
        "Model": label,
        "Brier score": round(brier_score_loss(y_test, prob), 4),
        "ECE": round(expected_calibration_error(y_test, prob), 4),
    })
plt.plot([0, 1], [0, 1], linestyle="--", color="gray", label="Perfectly calibrated")
plt.xlabel("Mean predicted probability")
plt.ylabel("Observed fraction with cardiovascular disease")
plt.title("Calibration Curve (Reliability Diagram)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

calibration_table = pd.DataFrame(calib_rows)
display(calibration_table)

# Use the calibrated model only if it improves the Brier score
if brier_score_loss(y_test, y_prob_cal) < brier_score_loss(y_test, y_prob_final):
    deploy_model, deploy_name, y_prob_deploy = calibrated_model, f"{final_name} + isotonic calibration", y_prob_cal
else:
    deploy_model, deploy_name, y_prob_deploy = final_pipeline, final_name, y_prob_final

print(f"Model used for the app: {deploy_name}")
```

## Decision Threshold for Screening

A fixed 0.5 threshold misses about one third of the patients. For a screening tool, missing a patient is worse than a false alarm, so the threshold is chosen for a **target recall of 80%**.

To avoid tuning on the test set, the threshold is derived from **out-of-fold predictions on the training set**, and only then evaluated on the test set. Higher recall always comes with more false alarms; the trade-off table shows how much.

```python
# ==========================================
# STEP 15: Threshold Selection (target recall)
# ==========================================

from sklearn.model_selection import cross_val_predict
from sklearn.metrics import ConfusionMatrixDisplay

TARGET_RECALL = 0.80

# Out-of-fold probabilities on the TRAINING set (no test data involved)
oof_prob = cross_val_predict(
    clone(deploy_model), X_train, y_train,
    cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    method="predict_proba"
)[:, 1]

prec_oof, rec_oof, thr_oof = precision_recall_curve(y_train, oof_prob)

def threshold_for_recall(target):
    return float(thr_oof[rec_oof[:-1] >= target].max())

THRESHOLD = threshold_for_recall(TARGET_RECALL)
print(f"Selected threshold for {TARGET_RECALL:.0%} recall: {THRESHOLD:.3f}")

# Precision / recall as a function of the threshold
plt.figure(figsize=(9, 5))
plt.plot(thr_oof, prec_oof[:-1], label="Precision")
plt.plot(thr_oof, rec_oof[:-1], label="Recall")
plt.axvline(THRESHOLD, linestyle="--", color="gray", label=f"Chosen threshold = {THRESHOLD:.2f}")
plt.axvline(0.5, linestyle=":", color="black", label="Default threshold = 0.50")
plt.xlabel("Decision threshold")
plt.ylabel("Score")
plt.title("Precision and Recall vs Threshold (out-of-fold, training set)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Confusion matrices on the test set: default vs chosen threshold
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
for ax, thr, title in [(axes[0], 0.5, "Threshold = 0.50"),
                       (axes[1], THRESHOLD, f"Threshold = {THRESHOLD:.2f}")]:
    ConfusionMatrixDisplay.from_predictions(
        y_test, (y_prob_deploy >= thr).astype(int), ax=ax,
        display_labels=["No CVD", "CVD"], values_format="d", colorbar=False
    )
    ax.set_title(title)
plt.suptitle(f"{deploy_name}: Test-Set Confusion Matrix")
plt.tight_layout()
plt.show()

# Trade-off table (thresholds from out-of-fold data, results on the test set)
tradeoff_rows = []
for label, thr in [("default 0.5", 0.5)] + [(f"target recall {t:.0%}", threshold_for_recall(t))
                                            for t in [0.70, 0.75, 0.80, 0.85, 0.90]]:
    pred = (y_prob_deploy >= thr).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, pred).ravel()
    tradeoff_rows.append({
        "Setting": label,
        "Threshold": round(thr, 3),
        "Recall": round(tp / (tp + fn), 3),
        "Precision": round(tp / (tp + fp), 3),
        "Specificity": round(tn / (tn + fp), 3),
        "Missed patients (FN)": int(fn),
        "False alarms (FP)": int(fp),
    })
tradeoff_df = pd.DataFrame(tradeoff_rows)
print("=== Threshold trade-off on the test set ===")
display(tradeoff_df)
```

## Explainability (SHAP-style Contributions)

XGBoost's built-in SHAP values (`pred_contribs`) are used, so no extra library is required for the main plots. One-hot columns are **grouped back into their original feature** (for example `smoke`, `cholesterol`), which fixes the split "Smoker" / "Non-Smoker" rows of the raw importance chart. Values are in log-odds of the uncalibrated XGBoost model: positive values push the prediction towards cardiovascular disease.

```python
# ==========================================
# STEP 17: SHAP Contributions (global + single patient)
# ==========================================

import xgboost as xgb

fitted_pre = final_pipeline.named_steps["preprocessor"]
booster = final_pipeline.named_steps["model"].get_booster()
feature_names_out = list(fitted_pre.get_feature_names_out())

def original_feature(transformed_name):
    for f in numeric_features:
        if transformed_name == f"num__{f}":
            return f
    for f in categorical_features:
        if transformed_name.startswith(f"cat__{f}_"):
            return f
    return transformed_name

orig_names = [original_feature(n) for n in feature_names_out]

def shap_contributions(X_frame):
    """Return transformed matrix, per-feature SHAP values and the bias term."""
    Xt = fitted_pre.transform(X_frame)
    contribs = booster.predict(xgb.DMatrix(Xt), pred_contribs=True)
    return Xt, contribs[:, :-1], contribs[:, -1]

def group_by_original(values):
    return pd.DataFrame(values, columns=orig_names).T.groupby(level=0).sum().T

# ---- Global importance (mean |SHAP|, grouped by original feature)
X_shap = X_test.sample(n=min(3000, len(X_test)), random_state=42)
Xt_shap, sv_shap, _ = shap_contributions(X_shap)

mean_abs = group_by_original(sv_shap).abs().mean().sort_values()

plt.figure(figsize=(9, 5.5))
plt.barh(mean_abs.index, mean_abs.values, color="#2A9D8F")
plt.xlabel("Mean |SHAP value| (log-odds)")
plt.title("Global Feature Importance (SHAP, one-hot columns grouped)")
plt.tight_layout()
plt.show()

# ---- Optional beeswarm plot (needs: pip install shap)
try:
    import shap
    plain_names = [n.replace("num__", "").replace("cat__", "") for n in feature_names_out]
    shap.summary_plot(sv_shap, Xt_shap, feature_names=plain_names, max_display=15, show=False)
    plt.title("SHAP Summary (beeswarm)")
    plt.tight_layout()
    plt.show()
except ImportError:
    print("Beeswarm plot skipped: install the 'shap' package (pip install shap) to enable it.")

# ---- Single-patient explanation
EXAMPLE_PATIENT = {
    "age_years": 55, "gender": 1, "height": 165, "weight": 75,
    "ap_hi": 140, "ap_lo": 85, "cholesterol": 2, "gluc": 1,
    "smoke": 0, "alco": 0, "active": 1,
}
example_df = pd.DataFrame([EXAMPLE_PATIENT])[X_train.columns]

_, sv_ex, bias_ex = shap_contributions(example_df)
contrib_ex = group_by_original(sv_ex).iloc[0]
contrib_ex = contrib_ex.reindex(contrib_ex.abs().sort_values().index)

plt.figure(figsize=(9, 5))
plt.barh(
    contrib_ex.index, contrib_ex.values,
    color=["#E76F51" if v > 0 else "#2A9D8F" for v in contrib_ex.values]
)
plt.axvline(0, color="black", linewidth=0.8)
plt.xlabel("Contribution to risk (log-odds); red = raises risk, teal = lowers risk")
plt.title("Why this prediction? (example patient)")
plt.tight_layout()
plt.show()

raw_prob = 1 / (1 + np.exp(-(bias_ex[0] + sv_ex.sum())))
print(f"Uncalibrated XGBoost probability for the example patient: {raw_prob:.2%}")
```

```python
# ==========================================
# STEP 19: Prediction Function (with input validation)
# ==========================================

INPUT_RANGES = {
    "age_years": (AGE_MIN, AGE_MAX),
    "height": (HEIGHT_MIN, HEIGHT_MAX),
    "weight": (WEIGHT_MIN, WEIGHT_MAX),
    "ap_hi": (SBP_MIN, SBP_MAX),
    "ap_lo": (DBP_MIN, DBP_MAX),
}

ALLOWED_CATEGORIES = {
    "gender": {1, 2},
    "cholesterol": {1, 2, 3},
    "gluc": {1, 2, 3},
    "smoke": {0, 1},
    "alco": {0, 1},
    "active": {0, 1},
}

def predict_cardio_risk(
    age_years, gender, height, weight, ap_hi, ap_lo,
    cholesterol, gluc, smoke, alco, active
):
    """
    Predict cardiovascular disease risk for a single person.
    Inputs outside the range seen in training are rejected, because the
    model has no experience with them.
    """
    values = {
        "age_years": age_years, "gender": gender, "height": height, "weight": weight,
        "ap_hi": ap_hi, "ap_lo": ap_lo, "cholesterol": cholesterol, "gluc": gluc,
        "smoke": smoke, "alco": alco, "active": active,
    }

    for name, (lo, hi) in INPUT_RANGES.items():
        if not lo <= values[name] <= hi:
            raise ValueError(f"{name}={values[name]} is outside the supported range {lo}-{hi}")
    for name, allowed in ALLOWED_CATEGORIES.items():
        if values[name] not in allowed:
            raise ValueError(f"{name}={values[name]} must be one of {sorted(allowed)}")
    if ap_hi <= ap_lo:
        raise ValueError("Systolic blood pressure must be higher than diastolic blood pressure")

    input_data = pd.DataFrame([values])[X_train.columns]

    probability = float(deploy_model.predict_proba(input_data)[0, 1])
    prediction = int(probability >= THRESHOLD)   # screening threshold, not 0.5

    return {
        "prediction": prediction,
        "risk_probability": probability,
        "threshold": THRESHOLD,
    }

print("Prediction function created successfully.")
```
